```python
# =========================================================
#          GAME ALTER EGO (GOOGLE COLAB VERSION)
#             TUGAS KELOMPOK PEMROGRAMAN
# =========================================================

# Variabel Warna ANSI yang Kontras untuk Background Putih Google Colab
TEBAL   = "\033[1m"
ORANYE  = "\033[33m"  # Lebih kelihatan di layar putih dibanding kuning terang
HIJAU   = "\033[32m"
UNGU    = "\033[35m"  # Menggantikan biru muda agar lebih kontras
MERAH   = "\033[31m"
RESET   = "\033[0m"

LEBAR_LAYAR = 43

# --- STEP 1: HALAMAN PEMBUKA & INPUT DATA ---
print(TEBAL + ORANYE + "===========================================" + RESET)
print(TEBAL + ORANYE + "     SELAMAT DATANG DI GAME ALTER EGO      ".center(LEBAR_LAYAR) + RESET)
print(UNGU + "  Temukan Sisi Lain yang Kamu Sembunyikan  ".center(LEBAR_LAYAR) + RESET)
print(TEBAL + ORANYE + "===========================================" + RESET)
print()

# Mengambil data awal dari pemain (Di Colab akan muncul kotak input di bawah)
nama = input(TEBAL + "Masukkan nama atau inisialmu : " + RESET)
umur = input(TEBAL + "Berapa umurmu sekarang?      : " + RESET)

print()
print(HIJAU + f"Halo {nama}! Umurmu {umur} tahun ya.".center(LEBAR_LAYAR) + RESET)
print(UNGU + "Persiapkan dirimu, game akan dimulai...".center(LEBAR_LAYAR) + RESET)
print(TEBAL + ORANYE + "===========================================" + RESET)


# --- STEP 2: SISTEM POIN ---
poin_thinker = 0
poin_adventurer = 0


# --- STEP 3: DAFTAR PERTANYAAN KEPRIBADIAN ---
print("\n" + TEBAL + UNGU + "--- PERTANYAAN 1 ---".center(LEBAR_LAYAR) + RESET)
print("Saat akhir pekan atau waktu luang,")
print("apa yang lebih kamu pilih?")
print(ORANYE + "A. Berdiam di kamar, baca buku/nonton." + RESET)
print(ORANYE + "B. Keluar rumah spontan, jalan-jalan." + RESET)
print("-" * LEBAR_LAYAR)

jawab1 = input(TEBAL + "Jawabanmu (A/B): " + RESET).upper()

if jawab1 == "A":
    poin_thinker += 1
elif jawab1 == "B":
    poin_adventurer += 1


print("\n" + TEBAL + UNGU + "--- PERTANYAAN 2 ---".center(LEBAR_LAYAR) + RESET)
print("Bagaimana caramu mengambil")
print("keputusan yang penting?")
print(ORANYE + "A. Dipikirkan matang-matang sendirian." + RESET)
print(ORANYE + "B. Ikut kata hati dan langsung bertindak." + RESET)
print("-" * LEBAR_LAYAR)

jawab2 = input(TEBAL + "Jawabanmu (A/B): " + RESET).upper()

if jawab2 == "A":
    poin_thinker += 1
elif jawab2 == "B":
    poin_adventurer += 1


# --- STEP 4: HALAMAN HASIL AKHIR ---
print("\n" + TEBAL + ORANYE + "===========================================" + RESET)
print(TEBAL + "          MENGHITUNG HASILMU...          ".center(LEBAR_LAYAR) + RESET)
print(TEBAL + ORANYE + "===========================================" + RESET)
print()

if poin_thinker > poin_adventurer:
    print(TEBAL + f"Halo {nama}, Alter Ego kamu adalah:" + RESET)
    print(TEBAL + MERAH + "🔮 THE SILENT THINKER (Pemikir Rahasia) 🔮".center(LEBAR_LAYAR) + RESET)
    print("-" * LEBAR_LAYAR)
    print("Di luar kamu mungkin terlihat biasa saja,")
    print("tapi di dalam kepalamu ada dunia luas.")
    print("Kamu pengamat hebat yang jenius!")
elif poin_adventurer > poin_thinker:
    print(TEBAL + f"Halo {nama}, Alter Ego kamu adalah:" + RESET)
    print(TEBAL + HIJAU + "🦅 THE WILD ADVENTURER (Petualang) 🦅".center(LEBAR_LAYAR) + RESET)
    print("-" * LEBAR_LAYAR)
    print("Kamu menyimpan jiwa kebebasan yang tinggi.")
    print("Jiwa alter egomu selalu haus tantangan")
    print("dan sangat berani di situasi tertentu!")
else:
    print(TEBAL + f"Halo {nama}, Alter Ego kamu adalah:" + RESET)
    print(TEBAL + UNGU + "⚖️ THE BALANCED SOUL (Jiwa Seimbang) ⚖️".center(LEBAR_LAYAR) + RESET)
    print("-" * LEBAR_LAYAR)
    print("Kamu adalah orang yang sangat seimbang.")
    print("Bisa menjadi sangat tenang, tapi juga")
    print("siap berpetualang di momen yang tepat!")

print(TEBAL + ORANYE + "===========================================" + RESET)


# --- STEP 5: SAMBUNGAN KE SCRATCH (OUTPUT TOKEN) ---
print("\n" + TEBAL + MERAH + "===========================================" + RESET)
print(TEBAL + MERAH + "          LANGKAH SELANJUTNYA            ".center(LEBAR_LAYAR) + RESET)
print(TEBAL + MERAH + "===========================================" + RESET)
print("Untuk melihat visualisasi karakter")
print("Alter Egomu secara menarik,")
print("ikuti langkah berikut:\n")
print(TEBAL + "1. Salin KODE RAHASIA di bawah ini:" + RESET)

# Membuat kode rahasia otomatis
kode_rahasia = f"EGO-{nama[:3].upper()}-{poin_thinker}{poin_adventurer}"
print(TEBAL + HIJAU + f"   >>> {kode_rahasia} <<< \n" + RESET)

print(TEBAL + "2. Buka link game Scratch kami:" + RESET)
print(UNGU + "   https://scratch.mit.edu/projects/1326733850" + RESET)
print(TEBAL + "\n3. Masukkan KODE RAHASIA tadi")
print("   di dalam game Scratch!" + RESET)
print(TEBAL + MERAH + "===========================================" + RESET)
print(TEBAL + HIJAU + "       Terima kasih sudah bermain!       ".center(LEBAR_LAYAR) + RESET)
print(TEBAL + MERAH + "===========================================" + RESET)
```
